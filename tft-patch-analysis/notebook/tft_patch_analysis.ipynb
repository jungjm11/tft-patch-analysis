```python
import pandas as pd

df = pd.read_csv("../data/patch_data.csv")
df.head()
```

```python
print(df.shape)
print(df["patch"].value_counts())
print(df.dtypes)
```

```python
before = df[df["patch"] == "16.1c"].copy()
after = df[df["patch"] == "16.2"].copy()

print("16.1c rows:", len(before))
print("16.2 rows:", len(after))
```

```python
before_sorted = before.sort_values("pick_rate", ascending=False)
after_sorted = after.sort_values("pick_rate", ascending=False)

before_sorted.head(10)
```

```python
before_best = before.sort_values("avg_placement", ascending=True)
after_best = after.sort_values("avg_placement", ascending=True)

before_best.head(10)
```

```python
before_comps = set(before["comp"])
after_comps = set(after["comp"])

removed_comps = before_comps - after_comps
new_comps = after_comps - before_comps

print("16.1c에 있었는데 16.2에서 사라진 덱:")
print(removed_comps)

print("\n16.2에서 새로 보이는 덱:")
print(new_comps)
```

```python
import matplotlib.pyplot as plt
```

```python
top10_before = before.sort_values("pick_rate", ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.barh(top10_before["comp"], top10_before["pick_rate"])
plt.xlabel("Pick Rate")
plt.ylabel("Comp")
plt.title("Top 10 Comps by Pick Rate - Patch 16.1c")
plt.gca().invert_yaxis()
plt.show()
```

```python
top10_after = after.sort_values("pick_rate", ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.barh(top10_after["comp"], top10_after["pick_rate"])
plt.xlabel("Pick Rate")
plt.ylabel("Comp")
plt.title("Top 10 Comps by Pick Rate - Patch 16.2")
plt.gca().invert_yaxis()
plt.show()
```

```python
df.groupby("patch")[["pick_rate", "avg_placement", "top4_rate", "win_rate"]].mean()
```

```python
df["comp"] = df["comp"].replace({
    "Void Baron": "Void Baron Nashor"
})
```

```python
before = df[df["patch"] == "16.1c"].copy()
after = df[df["patch"] == "16.2"].copy()
```

```python
df["comp"] = df["comp"].replace({
    "Ap Demacia": "AP Demacia"
})
```

```python
def top_n_concentration(data, n):
    return data.sort_values("pick_rate", ascending=False).head(n)["pick_rate"].sum()

print("Top 3 concentration")
print("16.1c:", top_n_concentration(before, 3))
print("16.2 :", top_n_concentration(after, 3))

print("\nTop 5 concentration")
print("16.1c:", top_n_concentration(before, 5))
print("16.2 :", top_n_concentration(after, 5))
```

```python
common_comps = sorted(set(before["comp"]).intersection(set(after["comp"])))
print(common_comps)
```

```python
comparison = (
    before[before["comp"].isin(common_comps)][["comp", "pick_rate", "avg_placement", "top4_rate", "win_rate"]]
    .merge(
        after[after["comp"].isin(common_comps)][["comp", "pick_rate", "avg_placement", "top4_rate", "win_rate"]],
        on="comp",
        suffixes=("_16_1c", "_16_2")
    )
)

comparison["pick_rate_change"] = comparison["pick_rate_16_2"] - comparison["pick_rate_16_1c"]
comparison["avg_placement_change"] = comparison["avg_placement_16_2"] - comparison["avg_placement_16_1c"]

comparison.sort_values("pick_rate_change")
```

```python
plot_df = comparison.sort_values("pick_rate_change")

plt.figure(figsize=(10,6))
plt.barh(plot_df["comp"], plot_df["pick_rate_change"])
plt.xlabel("Pick Rate Change (16.2 - 16.1c)")
plt.ylabel("Comp")
plt.title("Pick Rate Change for Shared Compositions")
plt.axvline(0, linestyle="--")
plt.show()
```

```python
comparison.sort_values("pick_rate_change").round(2)
```

```python
comparison.to_csv("../output/shared_comp_comparison.csv", index=False)
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

os.makedirs("../output", exist_ok=True)

concentration_df = pd.DataFrame({
    "patch": ["16.1c", "16.2"],
    "top3_concentration": [
        before.sort_values("pick_rate", ascending=False).head(3)["pick_rate"].sum(),
        after.sort_values("pick_rate", ascending=False).head(3)["pick_rate"].sum()
    ],
    "top5_concentration": [
        before.sort_values("pick_rate", ascending=False).head(5)["pick_rate"].sum(),
        after.sort_values("pick_rate", ascending=False).head(5)["pick_rate"].sum()
    ]
})

concentration_df.plot(x="patch", y=["top3_concentration", "top5_concentration"], kind="bar", figsize=(8,6))
plt.xlabel("Patch")
plt.ylabel("Concentration (Sum of Pick Rates)")
plt.title("Meta Concentration Comparison")
plt.xticks(rotation=0)

plt.tight_layout()
plt.savefig("../output/meta_concentration_comparison.png", dpi=300, bbox_inches="tight")
plt.show()
```
